**Common error**: using a *value* as if it were a *position*

In Lecture 09 we also filled a list by **position**, using a pre-allocated list (`[None] * n`). Watch what happens when we mix the two ideas up:

</div>

```python
precip_mm = [None] * len(mean_precip_in)     # 12 empty slots

for p in mean_precip_in:
    precip_mm[p] = p * 25.4                  # p is a VALUE like 4.02, not a position!
```

`p` holds the *values* in the list (4.02, 4.47, ...), not their positions (0, 1, ...). There's no such thing as "slot 4.02" in a list, so Python raises `TypeError: list indices must be integers or slices, not float`. (In Lecture 09 the same mix-up gave an `IndexError` instead: the cohesion values there were whole numbers like 0 and 4000, so Python tried to use them as positions, in a list that didn't have those slots.)

**The fix:** if you need positions, loop over positions. That's what `range()` is for.

## Part 2 : `range()`, and why we stopped counting with floats

Remember the `while` loop from Lecture 09 that counted Rw up by 0.1 and went one step too far? Here's the same idea, stripped down:

```python
x = 0.0
while x < 1.0:
    print(x)
    x += 0.1
```

We wanted `0.0` through `0.9`. Instead we got strange values like `0.30000000000000004` and a sneaky extra pass at `0.9999999999999999`. The computer stores numbers in binary, and `0.1` can't be stored exactly in binary (just like 1/3 can't be written exactly in decimal). Add it up ten times and the tiny errors pile up.

The fix from Lecture 10: **count with whole numbers, then divide.** `range(10)` gives exactly `0, 1, ..., 9`, and dividing fresh each time means no error builds up.

```python
for i in range(10):
    print(i / 10)
```

<div class="alert alert-success">

**Tip**: `range(n)` gives `0, 1, ..., n-1`, which is exactly the list of **positions** in a list of length `n`. That's why `range(len(my_list))` shows up so often.

</div>

## A helpful table

| **pattern** | **looks like** |
|:-|:-|
| loop over values | `for p in mean_precip_in:` |
| build a new list | `new = []` before the loop, `new.append(...)` inside it |
| loop over positions | `for i in range(len(months)):` |
| count safely | `range()` with whole numbers, then divide, never `x += 0.1` |
| walk parallel lists | `for month, p in zip(months, mean_precip_in):` |
| positions *and* values | `for i, month in enumerate(months):` |
| count / filter | a counter or list **before** the loop, an `if` **inside** it |
