Now let's use `np.array()` to turn our list into an array:

````python
week_highs = np.array(highs_list)

print(week_highs)
print(type(week_highs))
````

#### An array is like a list. 
It prints without commas and its type is `numpy.ndarray`: an "n-dimensional array". That's NumPy's name for an array.
##### There are two big differences between arrays and lists:
1. in an array, every element ahs the same type
2. in an array, math happens to every element at once

Arrays can tell you about themselves:

````python
print(len(week_highs))
print(week_highs.shape)
print(week_highs.size)
print(week_highs.dtype)
````

#### What just happened?
|**what you type**|**what it tells you**|**for `week_highs`**|
|:--|:--|:--|
|`len(week_highs)`|how many elements (same as for lists)| 7|
|`week_highs.shape`|the size along each dimension (as a tuple)| `(7,)`: one dimension, 7 long|
|`week_highs.size`|total number of elements| 7|
|`week_highs.dtype`|the **one** data type every element shares| `int64`: whole numbers|

Notice that `.shape`, `.size`, and `.dtype` have **no parentheses**. They are **attributes**: facts the array already knows about itself, not functions that have to run. `len()` is a regular function, so it does get parentheses.

#### "Every element ahs the same type": seeing it
A list can hold amix of types. An array can't, so NumPy **converts** everything to one type. Predict what each line prints, then run it:

````python
print(np.array([44, 49.5]))
print(np.array([44, 49.5]).dtype)
print(np.array([44, 49.5, "hot"]))
````

Mixing a whole number with a decimal turns **everything** into decimals (`float64`): note the `44.`. Add a string and **everything** becomes a string, `'44'` included. YOu can't do math on `'44'`, so keep your arrays to one kind of data.

## Part 2 : the same math on every element
Let's convert the week's highs to Celcius. You already know how to do this with loop and `.append()`:

````python
highs_c_list = []

for t in highs_list:
    highs_c_list.append((t-32)*(5/9))

print(high_c_list)
````

That work, and you can write it from scratch: that's three weeks of loop practice paying off. Now here's the NumPy version:

````python
week_highs_c = (week_highs - 32) * (5/9)

print(week_highs_c)
````

#### What just happened?
NumPy ran the loop **for you**. When you do math between an array and a single number, NumPy applies to **every element**, in order, and hands back a **new array** the same length. This is called **element-wise**math (you'll also hear **vectorized**).
|**position**| **`week_highs[i]` ($^\circ$F)**|**`(week_highs[i] - 32) * (5/9)`**|**`week_highs_c[i]`**|
|--:|--:|:--:|--:|
|0|44|(44 - 32) * 5/9| 6.67|
|1|49|(49 - 32) * 5/9| 9.44|
|1|44|(44 - 32) * 5/9| 6.67|
|..|..|..| 6.67|
|6|32|(32 - 32) * 5/9| 0.00|

It's the same loop you wrote above. It just runs inside NumPy, where it's shorter to type and *much* faster. `week_highs` itself doesn't change. This works for `+`, `-`, `*`, `/`, and `**`.

**Your own functions work on arrays too**, as long as they only do arithmetic inside:

````python
def f_to_c(temp_f):
    """Convert a temperature (or an array of temperatures) from Fahrenheit to Celsius."""
    return (temp_f - 32) * (5/9)

print(f_to_c(212))
print(f_to_c(week_highs))
````

#### Careful: lists do *not* work this way
Before you run the next two cells, **predict** what each line will print.

````python
print(highs_list * 2)
print(highs_list + highs_list)
````

````python
highs_list - 32
````

Lists treat `*` and `+` as *list* operations: `*2` **repeats the list and `+` **sticks two lists together**. And subtracting a number from a list doesn't mean anything at all, so Python raises a `TypeError`. Read the last line of the error: *unsupported operand type(s) for -: `list` and `int`.

The same symbols on the array mean math:

````python
print(week_highs * 2)
print(week_highs + week_highs)
````

|**expression**|**if it's a list**|**if it's an array**|
|:-|:-|:-|
|`x * 2`| repeats the list (14 elements)| doubles every element (7 elements)|
|`x + x`| joins the lists end to end (14 elements)| adds element by element (7 elements)|
|`x - 32`| `TypeError`| subtracts 32 from every element|


<div class="alert alert-danger">

**Common Error:** doing math on a list when you meant an array. If you get a `TypeError` about `'list'`, or a result that's twice as long as you expedcted, check `type()` and wrape your list in `np.array()`
    
</div>

#### Two arrays at once
When **both** sides are arrays of the **same length**, NumPy pairs them up **by position**, just like `zip()` did for your parallel lists. NOAA computes a day's **average temperature** as the average of it's high and low (Arguez et al., 2012). You made `week_lows` in Exercise 1, so:

````python
week_lows = np.array([28, 25, 30, 29, 34, 27, 27])

week_mean = (week_highs + week_lows) /2

print(week_highs)
print(week_lows)
print(week_mean)
````

Position 0 of `week_highs` goes with position 0 of `week_lows`, position 1 with position 1, and so on. No index variable, no `range(len(...))`

## Part 3 : `math` vs `np` - functions that work on arrays
Warm air can hold more water vapor than cold air. The **saturation vapor pressure**, $e_s$ is the most water vapor the air can hold before it starts to condense, and it depends only on temperature. A widely used formulat (Bolton, 2980) is:
$$e_s = 6.112 \, \exp\left(\frac{17.67\,T}{T + 243.5}\right)$$
where $T$ is temperature in $^\circ$C and $e_s$ is in hectopascals (hPa). the $\exp()$ means $e$ raised to that power. You've used `math.exp()` before, so let's write it as a function:

````python
import math

def sat_vapor_pressure(temp_c):
    """Saturation vapor pressure (hPa) at temperature temp_c (c), from Bolton (1980)."""
    return 6.112 * math.exp(17.67 * temp_c / (temp_c + 243.5))

print(sat_vapor_pressure(20))
````

That works for one temperature. Now let's hand it the whole week of highs:

````python
sat_vapor_pressure(week_highs_c)
````

#### Reading the error
Start at the **bottom**: it's a `TypeError`, and the message says that only a single number (a "0-dimensional array") can be converted to a Python scalar. Now look up at the arrow: the problem is inside `math.exp()`.

The `math` module was written for **one number at a time**. It has no diea what to do with seven of them.

The fix: **NumPy has its own versions** of the math functions, and the work element-wise.
|`math` module (one number)|NumPy (one number *or* a whole array)|
|:-|:-|
|`math.exp(x)`|`np.exp(x)`|
|`math.sqrt(x)`|`np.sqrt(x)`|
|`math.log(x)`|`np.log(x)`|
|`math.sin(x)`|`np.sin(x)`|
|`math.pi`|`np.pi`|


Swap one word and redefine the function:

````python
def sat_vapor_pressure(temp_c):
    """Saturation vapor pressure (hPa) at temperature temp_c (c), from Bolton (1980)."""
    return 6.112 * np.exp(17.67 * temp_c / (temp_c + 243.5))

sat_vapor_pressure(week_highs_c)
````

## Part 4 : making new arrays
So far we have built arrays from lists. NumPy can also build them from scratch:
|**what to write**|**what it makes**|**example**|**result**|
|:-|:-|:-|:-|
|`np.array(a_list)`| an array from a list| `np.array([44, 49, 44])`| `[44 49 44]`|
|`np.zeros(n)`| `n` zeros| `np.zeros(3)`| `[0. 0. 0.]`|
|`np.ones(n)`| `n` ones| `np.ones(3)`| `[1. 1. 1.]`|
|`np.arange(start, stop, step)`| like `range()`, but an array| `np.arange(1, 6)`| `[1 2 3 4 5]`|
|`np.linspace(start, stop, num)`| `num` evenly spaced values, **both ends included**| `np.linspace(0, 1, 5)`| `[0. 0.25 0.5 0.75 1.]`|

`np.arange()` follows the same rules as `range()`: the stop is **not** included. SO the day numbers for our first week are:

````python
week_days = np.arange(1,8)
print(week_days)
````

#### Remember the 1.0 that wasn't?
In Lecture 09, `while rw < 1.0: rw +=1` printed an extra `Rw = 1.0", because 0.1 can't be stored exactly in binary and the tiny errors piled up. Here are two ways to build 0.0, 0.1, ..., 1.0 with NumPy:

````python
print(np.arange(0, 1.0, 0.1))
print(np.linspace(0, 1, 11))
````

- `np.arange()` stops *before* 1.0, just like `range()` would. 
- `np.linspace()` asks a different question: not "what step>" but "**how many values** between these two ends?"
    - It calculates each value directly from the endpoints instead of adding 0.1 over and over so the ends come out exact.
 
<div class="alert alert-success">

**Tip**:
- **Want whole number steps?** | `np.arange()`
- **Want decimal spacing?** | `np.linspace()`

NumPy's own documentation reccomends `linspace` whenever the step isn't a whole number, for exactly the rounding reason we saw in Lecutre 09.
    
</div>

For example, Lab 03's water-table ratios, $R_w$ = 0.0, 0.2, ..., 1.0 are six values from 0 to 1

````python
rw = np.linspace(0, 1, 6)
print(rw)
````

|**column**| **what it holds**|**our variable**|

## Part 6 : one number for 365 - was 2025 normal?

Since Week 4, you've added things up with an accumulator: `total = 0` befoe a loop and `total += ...` inside it. Arrays have that loop built in:

````python
total_precip = prcp.sum()
print(total_precip)
````

Those trailing digits are a consequence of the fact that decimals aren't stored exactly in binary, and adding 265 of them leaves a tiny crumb. The eal answer is 54.91 in. **Format when you print** and the crumb disappears:

````python
normal_precip = 50.51

print(f"2025 total:   {total_precip:.2f} in")
print(f"Normal total: {normal_precip:.2f} in")
print(f"Difference:   {total_precip - normal_precip:+.2f} in")
````

*The normal comes from adding the 12 month precipitaiton normals in NOOA's 1991-202 US Climate Normals for Nashville.*

##### 2025 was about **4.4 in wetter** than normal. What about temperature?
|**what to write**|**what it gives you**|
|--|--|
|`a.sum()`| the total of every element|
|`a.mean()`| the average|
|`a.max()`| the biggest value|
|`a.min()`| the smallest value|

Each one also comes as a function: `np.sum(a)`, `np.mean(a)`, and so on do exactly the same thing.

````python
normal_tmean = 60.8 # F the average of the 12 monthly 1991-2020 tmean normals for this station

print(f"Hottest high: {tmax.max()} F")
print(f"Coldest cold: {tmin.min()} F")
print(f"2025 average temperature: {tmean.mean():.1f} F (normal {normal_tmean} F)")
````

So 2025 ran about **2 $^\circ$F warmer** and about **4 inches wetter** than the 1991-2020 normal, at least at this one station.

### Bonus: degree days

Energy planners measure heating and cooling demand in **degree days**, measured from a base of **65 °F**. Each day's **cooling degree days** are how far its average temperature is *above* 65 °F, and zero on days at or below 65 °F. Heating degree days are the same idea, *below* 65 °F.

`np.maximum(a, 0)` compares every element of `a` to 0 and keeps whichever is bigger, so it turns negative numbers into 0. Use it with `tmean` to calculate Nashville's total **cooling** and **heating** degree days for 2025. 

**Was 2025 a heating year or a cooling year?**


````python
# your answer here
````

<details>
<summary style="color:green;"><strong>Answer.</strong></summary>

```python
cdd = np.maximum(tmean - 65, 0)
hdd = np.maximum(65 - tmean, 0)

print(f"Cooling degree days: {cdd.sum():.0f}")
print(f"Heating degree days: {hdd.sum():.0f}")
```
---
```
Cooling degree days: 2197
Heating degree days: 3034
```

Even in a warm year, Nashville needed more heating than cooling, measured this way. The trick is that `tmean - 65` is negative on cool days, and `np.maximum(..., 0)` replaces those negatives with zeros before we add them up.
