To make that table an array, give `np.array()` **a list of rows**, where each row is itself a list. You've seen this shape before: it's a nested list, from Lecture 04.


````python
small = np.array([[39.6, 43.4, 51.5],
                 [31.7, 32.7, 41.6]])

print(small)
print("shape:", small.shape)
print("dimensions:", small.ndim)
print("size:", small.size)
````

#### What just happened?

|**what you type**|**what it tells you**|**for `small`**|
|:--|:--|:--|
|`small.shape`|size along each dimension, **rows first, then columns**|`(2, 3)`: 2 rows, 3 columns|
|`small.ndim`|how many dimensions|`2`|
|`small.size`|total number of elements|`6` (2 × 3)|

`ndim` is now, but it's just counting how many numbers are in `.shape`. Your ID arrays from last week had `.ndim` of 1.

#### Indexing: `[row, col]`
One index inside the brackets gets you a **row**. Two indices, separated by a comma, get you **one value**: row first, then column.

````python
print("row 1:", small[1])
print("row 1, column 2:", small[1, 2])
print("every row, column 0:", small[:, 0])
````

That last line is exactly the move you used earlier, `tmax = data[:, 2]. The `:` means "every row".

|**you write**|**read it as**|**you get**|
|:--|:--|:--|
|`small[1]`|row 1|`[31.7 32.7 41.6]`|
|`small[1, 2]`|row 1, column 2|`41.6`|
|`small[:, 0]`|every row, column 0|`[39.6 31.7]`|
|`small[0, 1:3]`|row 0, columns 1 and 2|`[43.4 51.5]`|

<div class = "alert alert-danger">

**Common Error:** `small[1][2]` also works (its how you indexed nested lists), but get in the habit of `small[1, 2]`. The comma version is what makes slices like `small[:, 0]` possible. `small[:][0]` does **not** give you column 0: it gives you row 0.
    
</div>

## Part 2 : `axis` - summarizing in one direction

Last week, `.mean()` boiled 365 numbers down to one. What happens on a table?

````python
print(tavg.mean())
````

That's the average of alllll 71 numbers, mixing Phoenix's July with Utquiġvik's February. It's correct, but it doesn't mean anything...What we want is **one average per station** or **one average per month**. The `axis` argument picks the direction:

````python
annual = tavg.mean(axis=1)
by_month = tavg.mean(axis=0)

print(annual.shape)
print(by_month.shape)
````

#### What just happened?

The numbers in `axis` are the same positions as in `.shape`. `tavg.shape` is `(6, 12)`, so **axis 0 is the rows** (6 stations) and **axis 1 is the columns** (12 months).

`axis` tells NumPy which one to **use all off** when it averages:

- `axis=0` : average **all rows** together. Every station goes into each average, one month at a time, so you get **one value per month**.
    - averages down each column
- `aix=1` : average **all the columns** together. Every month goes into each average, one station at a time, so you get **one value per station**.
    - averages across each row
 
```
                 Jan    Feb    ...    Dec
  Nashville  [  39.6   43.4   ...   42.7 ]  ──▶  axis=1: all 12 months of this row
  Denver     [  31.7   32.7   ...   31.2 ]  ──▶        → one value per station, shape (6,)
  ...
  Utqiagvik  [ -11.5  -11.9   ...   -6.3 ]  ──▶
                  │      │             │
                  ▼      ▼             ▼
            axis=0: all 6 rows of this column
                  → one value per month, shape (12,)
```

|**you write**|**averages together**|**you get one value per**|**shape**|**means**|
|:--|:--|:--|:--|:--|
|`tavg.mean(axis=0)`|all the **rows** (stations)|month|`(12,)`|each **month's** average across stations|
|`tavg.mean(axis=1)`|all the **columns** (months)|station|`(6,)`|each **station's** annual average|

The part that trips everyone up: `axis=0` means "rows," but you get one answer per **coulmn**. That's because the rows are what got used up. Check it with the shapes: `(6, 12)` with axis=0 used up leaves `(12,)`.

Arrays and loops work together: NumPy does the math, a loop does the labels

````python
for name, t in zip(stations, annual):
    print(f"{name:12s} {t:5.1f} F")
````

<div class = "alert alert-success">

**Lining up columns:** in `{name:12s}`, the `12s` means "print this **s**tring, padded with spaces to **12** characters wide." In `{t:5.1f}`, the `5` is a minimum width and `.1f` is one decimal place, as before. Widths don't change the values. They only make the output line up in neat columns.
    
</div>

Nashville's 60.8 $^\circ$F is the same annual normal you used last week. `axis` works on **every** summary method: `.sum()`, `.max()`, `.min()`, `.argmax()` too. Which month is the warmest at each station?

````python
warmest = tavg.argmax(axis=1)
print(warmest)

for name, i in zip(stations, warmest):
    print(f"{name:12s} warmest in {months[i]}")
````

Seattle peaks in **August**, not July (67.4 vs 67,1 $^\circ$F). Seattle sits next to the Pacific Ocean and Puget Sound, and water warms up slowly, so the coast's warmest weather lags into late summer.

That ocean effect shows up even more clearly i the **seasonal range**, the warmest month minus the coldest month:

````python
seasonal_range = tavg.max(axis=1) - tavg.min(axis=1)

for name, r in zip(stations, seasonal_range):
    print(f"{name:12s} {r:5.1f} F")
````

## Part 3 : broadcasting - math between arrays of different shapes

#### Two shapes you already know work
Math between an array and **one number** happens to every element. That's still true in 2D, so `f_to_c()` from earlier converts the whole table at once:

````python
def f_to_c(temp_f):
    """convert a temperature (or an array of temperatures) from F to C."""
    temp_c = (temp_f - 32) * (5/9)
    return temp_c

tavg_c = f_to_c(tavg)
print(tavg_c.shape)
print(tavg_c[3].round(1))
````

And math between two arrays of the **same shape** pairs them up position by position. In Lecture 15, Exercise 2 showed the third case: two **different** lengths fail.

`ValueError: operands could not be broadcast together with shapes (7,) (5,)`

That word **broadcast** is today's topic: the rules NumPy uses to decide whether two different shapes can work together.

#### A question that needs two shapes

Climate scientists often look at each month as **departure** from the station's annual average: how much warmer or colder than usual is this month **for this place**? We have the table, `(6, 12)`, and the annual averages, `(6,)`. Let's subtract:

````python
departure = tavg - annual
````

#### Reading the error

`shapes (6, 12) (6,)`. NumPy compares shapes by **lining them up from the right**:

```
tavg:     (6, 12)
annual:      (6,)
              ↑
        12 vs 6: different sizes, and neither is 1 → error
```

**The broadcasting rule:** line the shapes up from the right. Each pair of sizes must be **equal**, or one of them must be **1**. A size of 1 gets **stretched** to match.

So we need `annual` to be **column**, with shape `(6, 1)`. `np.newaxis` adds a new dimension of size 1 wherever you put it:

````python
annual_col = annual[:, np.newaxis]

print("Previously, the shape was:", annual.shape)
print("Now it's                 :", annual_col.shape)
print(annual_col)
````

Now when we line them up again:
```
tavg:     (6, 12)
annual:   (6,  1)
           ↑   ↑
           │   12 vs 1 → stretch the 1 across all 12 months ✓
           6 vs 6 → equal ✓
```

````python
departure = tavg - annual_col

print(departure.shape)
print(stations[3], departure[3].round(1))
print(stations[4], departure[4].round(1))
````

#### What just happened?

NumPy acted as if each station's annual average were copied across all 12 months, then sutracted element by element:

|**row**|**`tavg[row]`**|**minus (stretched)**|**`departure[row]`**|
|:--|:--|:--|:--|
|0 Nashville|`[39.6, 43.4, ..., 42.7]`|`[60.8, 60.8, ..., 60.8]`|`[-21.2, -17.4, ..., -18.1]`|
|3 Seattle|`[42.8, 44.0, ..., 42.0]`|`[53.7, 53.7, ..., 53.7]`|`[-10.9, -9.7, ..., -11.7]`|
|...|...|...|...|

It doesn't really make 12 copies (that would waste memory), but the result is the same. In Seattle, July runs about 13 $^\circ$F above the annual average. In Minneapiolis it's about 27 $^\circ$F below.

#### The other direction already works
What about subtracting each **month's** six-station average, `by_month`, with shape `(12,)`?
```
tavg:     (6, 12)
by_month:    (12,)
              ↑
        12 vs 12 → equal ✓, and the missing dimension on the left 
        counts as 1 → stretched down all 6 rows ✓
```

````python
vs_group = tavg - by_month

print(vs_group.shape)
print(vs_group[:,0].round(1))
````

No `np.newaxis` needed, because the 12s already line up on the right. In January, Phoenix ran 27.5 $^\circ$F warmer than the six-station average and Utqiaġvik 40.8 $^\circ$F colder.

|**shapes**|**line up from the right**|**works?**|
|:--|:--|:--|
|`(6, 12)` and a single number|the number stretches everywhere|✓|
|`(6, 12)` and `(6, 12)`|everything matches|✓|
|`(6, 12)` and `(12,)`|12 = 12, then the row stretches down|✓|
|`(6, 12)` and `(6,)`|12 ≠ 6|✗|
|`(6, 12)` and `(6, 1)`|12 vs 1 stretches, 6 = 6|✓|
|`(7,)` and `(5,)`|7 ≠ 5|✗|

<div class="alert alert-success">

**Tip:** when broadcasting surprises you, **print the `.shape` of both sides** and line them up from the right on paper. Almost every broadcasting error is a row that should have been a column: fix it with `[:, np.newaxis]`.
    
</div>

## A useful table

|**1D (last week)**|**2D (today)**|
|:--|:--|
|`a.shape` is `(365,)`|`a.shape` is `(rows, cols)`, e.g. `(6, 12)`|
|`a[3]` is one value|`a[3]` is a whole **row**; `a[3, 6]` is one value|
|`a[5:8]` slices values|`a[:, 6]` is a column; `a[4, 5:8]` slices a row|
|`a.mean()` gives one number|`a.mean(axis=1)` is one per **row**; `a.mean(axis=0)` is one per **column**|
|same-length arrays pair up|shapes line up **from the right**; sizes must match or be 1|
|*(n/a)*|`b[:, np.newaxis]` turns `(6,)` into a column, `(6, 1)`|
