## II. Determine the rate of outgoing energy, per unit area of the earth's surface.</span>

Most of the solar flux is aborbed either in the atmosphere or by materials at Earth's surface. This absorbed solar energy is converted from infrared (IR) energy by the absorbing material. It is then emitted from Earth and radiated away to space in a form called **infrared radiation**. 

Thus, the outgoing energy takes 2 forms:
1. reflected solar energy
2. infrared radiation by objects that have absorbed such energy

The rate at which an object radiates energy away from itself depends on its temperature.

<center> <span style="color:red;font-size:1.9em;font-weight:800;">The hotter an object is, the more radiation it emits. <span></center>

#### <center>Objects that are *maximally effective radiators of energy* are called **blackbodies**. Such objects are also perfect absorbers.</center>

<center>In otherwords, blackbodies perfectly absorb all of the light that hits it and then perfectly emitt all of that light back out as energy. A perfect blackbody is in equilibrium and should sit at a steady temperature.</center>

### The <u>Stefan-Boltzmann Law</u> states that the energy given off by a blackbody is a function of its temperature:
$$ E = \sigma T^4 \tag {Eq. 2}$$

where $F$ is the radiant energy output (with units of energy per unit time per unit area), $T$ is the absolute temperature of the object (in kelvin, K), and $\sigma$ is the Stefan-Boltzmann constant:
$$\sigma = 5.67 \text{x}10^-8\space J/m^2 sec\space K^4$$

$\sigma$'s units can also be $W/(m^2K^4)$

### IV. Determine how much *doesn't* make it to Earth's surface to begin with.

One process we might be missing is **abledo**.

![albedo](../figures/albedo.jpg)

**Albedo** is the fraction of light that a surface reflects. If it is all reflected, the albedo is equal to 1. If 20% is relfected, the albedo is 0.2. The albedo of Earth's surface (which includes the amosphere, oceans, and all land surfaces) determines how much incoming solar energy is immediately reflected back into space.

Mathmatically, we can represent the amount of energy reflected due to albedo as:
$$a\frac{\Omega}{4} \tag{Eq. 6}$$
where $a$ is albedo represented as a fraction.

If we assume that we need to add albedo, we can re-write Eq. 2 for the total outgoing energy as:

$$E_{out} = a\frac{\Omega}{4} + \sigma T^4 \tag{Eq. 7}$$

#### <center>Which means that our flux in and out becomes...</center>
| Flux In | Flux Out |
|--------|----------|
| $\frac{\Omega}{4}$ | $a\frac{\Omega}{4} + \sigma T^4$ |

#### <center>and we set those two equations equal to each other again...</center>

$$\frac{\Omega}{4} = a\frac{\Omega}{4} + \sigma T^4$$

#### <center>...and simplifying so that our equation for temperature which includes albedo becomes...</center>

$$T^4 = \frac{(1-a)\Omega}{4\sigma} $$
<center>or, simplified: </center>

$$T = \sqrt[4]{\frac{(1-a)\Omega}{4\sigma}} \tag {Eq. 8}$$

## IIV. Determine how energy is being trapped.
***Click the arrow to the left of this box OR the box below that says `+12 cells hidden` expand and view Questions 14-17.***

In Question 13, you speculated about why Earth's actual temperature might differ from what a simple blackbody calculation predicts. One big piece of the answer is the **greenhouse effect**.

The Sun is very hot (around 5800 K), so most of its energy arrives at Earth as **shortwave radiation** (which is mostly visible and near-infrared light). Earth is much cooler (around 255 K), so it radiates its energy back to space as **longwave radiation** (infrared light at much longer wavelengths).

Gasses in the atmosphere don't absorb every wavelength equally. Some gasses (**greenhouse gasses**) are especially good at absorbing longwave infrared radiation. Other gases that make up most of the atmosphere like N$_2$ and O$_2$, barely abosrb any of it. When a greenhouse gas absorbs outgoing infrared radiation, it re-emits that energy in all directions, including back down towards Earth's surface. This is part of why the surface ends up warmer than the simple blackbody prediction from Question 9.

![figure](../figures/greenhouse_absorption_realdata.png)
<span style="font-size:0.85em;color:#6b7a8d;"><em>The orange curve shows the wavelengths of incoming sunlight (shortwave); the dark curve shows the wavelengths of outgoing radiation from Earth (longwave). Gray bands mark wavelengths where H$_2$O, CO$_2$, O$_3$, CH$_4$, or N$_2$O absorb radiation. The green band is the "atmospheric window," where very little absorption happens.</em></span>

In the figure above, the Sun's curve (orange) and the Earth's curve (black) barely overlap with the same set of gray bands (wavelengths absorbed by atmospheric gases). Of the gases shown, the following are greenhouse gasses: H$_2$O, CO$_2$, O$_3$, CH$_4$, N$_2$O. Most of the incoming sunlight arrives at short wavelengths, around 0.2-2 microns, where only a few narrow gray bands overlap. This means that most of that shortwave energy passes through the atmosphere without being absorbed. 

Earth's outgoing radiation (dark curve) is concentrated at much longer wavelengths, roughly 5-30 microns, right where most of the gray absorption bands sit. That's why the atmosphere lets sunlight in fairly easily, but traps a lot of the infrared energy trying to leave--aka the Greenhouse Effect!
