## Part 1 — Variables and Objects

### What is a variable?

A **variable** is a name that *points* to a value stored in the computer's memory. Think of it like a labeled jar — you can put something in it, look inside it, or swap out its contents later.

```python
distance_km = 347
```

Here, `distance_km` is the name, and `347` is the value it points to.

#### Because Python variables just point to various objects, there is not need to "declare" the variable, or even require the variable to always point to information of the same type.

We could have just as easilyt defined `distance_km` as a string (don't worry we'll learn about strings in a bit!)
```python
distance_km = '345 kilometers`
```

Run the following cell:

````python
# Run this cell
distance_km = 347
print(distance_km)
````

##### A consequence of the "variable as pointer" approach is that if you have two variable names pointing towards the same thing, then changing one will change the other as well.

````python
# Run this cell
x = [1, 2, 3]
y = x
````

We just created two variables `x` and `y` which both point towards the same object. Because of this, if we modify the lsit via one of its names, we'll see that the "other" list will be modified as well:

````python
print(y)
````

Let's now "append" (add) a value to the end of that list in `x` and see what happens to `y`:

````python
x.append(4) # appends 4 to the list pointed to by x
print(y)    # y's list is modified as well!
````

##### Note, that if we use "`=`" to assign another value to `x`, this will not affect the value of `y` --assignment is simply a change of what object the variable points towards.

````python
x = 'something else'
print(y)
````

| Rule | Example |

### Everything is an Object.
Python is an object-oriented programming language, and in Python, everything is an object.

#### What does this mean?
Earlier, we saw that variables are simply pointers, and the variable names themselves have no attached `type` information. However, when you assign a value that inherently has a `type`, you can easily see that type using the function `type()`:

````python
x = 4
print('x is type', type(x))

y = 'hello'
print('y is type', type(y))

z = 3.14159
print('z is type', type(z))
````

We'll learn more about "types" later, but for now, I want you to notice that these **types are linked not to the variable names, but to the objects themselves.**

In object-oriented programming languages like Python, **an *object* is an entity that contains data along with associated metadata and/or functionality.** In Python everything is an ojbect, which means that every entity has some metadata (called *attributes* and associated functionality (called *methods*).

## Part 2 — Operators

In Python there are arithmetic, bitwise, and assignment operators. We're going to just focus on *arithemetic* and *assignment* operators in this class.

### Arithmetic Operations
An **operator** tells Python to perform a given operation on two **operands**
```
operand  operator  operand
   3         +        4
```
In python..
| Operator     | Name           | Description                                            |
|--------------|----------------|--------------------------------------------------------|
| ``a + b``    | Addition       | Sum of ``a`` and ``b``                                 |
| ``a - b``    | Subtraction    | Difference of ``a`` and ``b``                          |
| ``a * b``    | Multiplication | Product of ``a`` and ``b``                             |
| ``a / b``    | True division  | Quotient of ``a`` and ``b``                            |
| ``a // b``   | Floor division | Quotient of ``a`` and ``b``, removing fractional parts |
| ``a % b``    | Modulus        | Integer remainder after division of ``a`` by ``b``     |
| ``a ** b``   | Exponentiation | ``a`` raised to the power of ``b``                     |
| ``-a``       | Negation       | The negative of ``a``                                  |
| ``+a``       | Unary plus     | ``a`` unchanged (rarely used)                          |

You can use a programming language as a calculator. If you want to do that you will be using *arithmetic operators*.

- *Operators* define the type of calculation that is executed.
For example, the `+` operator

- Any operator works on *operands*. 
To extend the example below: the `+` operator adds the operands `a` and `b` together.

The application of these operators is intuitive. For example:

````python
(4 + 8) * (6.5 - 3)
````

The floor division and modulus operators might be new to you.

**Floor division** is true division with fractional parts truncated. So we could do true division like this:

````python
print(11 / 2) # true division
````

Which gives us `5.5`. If we wanted to do floor division, it would give us:

````python
print(11 // 2)
````

And we'd end up with just `5`.

The **modulus** or **modulo** operator will give you the remainder after division. So:

````python
print(11 % 2)
````

<div class="alert alert-danger">
    
**Note**: The `0.5` part of `5.5` when doing true division of 11 by 2 is referring to the fractional part, not the remainder when dividing 11 by 2.

</div>


The last tricky thing is that to raise something to a power (**exponentiation**), you need to use the double asteriks (`**`). This is easy to forget.

````python
v = 11**2
print("11^2 = ", v)
````

**A note about arithmetic operator precedence:**

Operator precedence defines the order in which operations are performed. 
<p><u>Here's what you need to know:</u>

1. Parentheses have the highest precedence, meaning that expressions inside parentheses are evaluated first.

2. Exponation `**` has precedence over multiplication `*`

3. Multiplication `*` has higher precedence than addition `+` and substraction `-`, and therefore multiplications are evaluated before additions.

4. Multiplication `*` is at the same *precendence level* as division `/`, and the statement is executed left to right. This is the same for addition and substraction.

### Assignment Operations

We've seen that variable can be assigned with the "`=`" operator, and the values stored for later use.

In Python, `=` is the **assignment operator** — it is *not* a statement of equality.

| What `=` does | Notes |
|---|---|
| Assigns the value on the right to the name on the left | The variable "gets" the value |
| Defines the variable if it doesn't already exist | No need to declare a type first |
| Overwrites the old value if the variable already exists | The previous value is gone |

<div class="alert alert-success">

**Tip:** Read `x = 5` as "x gets 5," not "x equals 5."


````python
a = 24
print(a)
````

We can add `2` to `a` by writing:

````python
a + 2
````

We might want to update the variable with this new value; in this case, we could combine the addition and the assignment and write
```python
a = a + 2
```
But Python is great and made this easier for us. There are built-in update operators for all of the arithemtic operations. So we can use those:

````python
a += 2 # equivalent to a = a + 2
print(a)
````

#### Built-in Update Operators
| operator | what it does |
|---|---|
| `a += b` | take value on right and add it to value on left |
| `a -= b` | take value on right and subtracit it from value on left |
| `a *= b` | take value on right and multiply it by value on left |
| `a /= b` | take value on left and divide it by value on right |
| `a //= b` | take value on left and do floor division by value on right |
| `a %= b` | take value on left figure out the reminder when divided by the value on the right |
| `a **= b` | take value on right and raise the value on left by it|

### Comparison operators 
Comparison operators compare two values and returns either `True` or `False`
| comparison operator | What it does | Example |
|---|---|---|
| `==` | is the value on the left equal to the value on the right | `age = 33` ; `age == 61` will return `False` |
| `!=` | is the value on the left UNequal to the value on the right | `age == 61` will return `True` |
| `>` | is the value on the left greater than the value on the right | `age > 61` will return `False` |
| `<` | is the value on the left less than the value on the right | `age < 61` will return `True` |
| `>=` | is the value on the left greater than or equal to the value on the right | `age >= 61` will return `False` |
| `>=` | is the value on the left less than or equal to the value on the right | `age <= 61` will return `True` |

Comparison operators will return a data type called a `boolean` variable or a `bool`
### `bool`s
A boolean variable can have only one of two possible values: `True` or `False`

## Reading An Error Message

Let's talk about that error message. We got a `ValueError` message when we tried to convert a string that wasn't made of numbers to a float. Let's recreate that by running the cell below again.


````python
planet_name_flt = float(planet)
print("Planet name as a float:", planet_name_flt)
````

#### **What happened?**

Python raised a `ValueError`. This error happesn when you try to convert something into a number, but the thing you're converting isn't actually a number. In this particular case, python has no idea how to turn the word "Mercury" inot a number, so it raises a `ValueError1 to tell you that it can't complete the conversion.

A few tips for reading error messages:
- Read the **last line first** — it usually names the type of error and gives the key detail
- The arrow (`---->`) points to the line where Python got stuck

#### Types of common errors:
| Error Type | What It Means | Example |
|---|---|---|
| `SyntaxError` | Code isn't valid Python "grammar" — happens before code runs | `if x = 5:` (should be `==`) |
| `NameError` | Using a variable/function name that doesn't exist | `print(totl)` when `totl` was never defined |
| `TypeError` | Wrong *type* of value for the operation — no value of that type would work | `"5" + 5` |
| `ValueError` | Right type, but an invalid *value* for that type | `int("hello")` |
| `IndexError` | Accessing a list/string/tuple position that doesn't exist | `my_list[10]` on a 3-item list |
| `KeyError` | Accessing a dictionary key that doesn't exist | `my_dict["b"]` when `"b"` isn't a key |
| `AttributeError` | Using a method/property that doesn't exist on that object | `"hello".append("!")` |
| `FileNotFoundError` | Trying to open a file that doesn't exist at that path | `open("myfile.txt")` |
| `ZeroDivisionError` | Dividing by zero | `10 / 0` |
| `ImportError` / `ModuleNotFoundError` | Trying to import a package/module Python can't find | `import pandsa` (misspelled) |

### Bonus: Cleaner printing with f-strings

So far we've printed variables like this, separating pieces with commas:

```python
print("The density of", mineral_name, "is", density, "g/cm3")
```

Python has a cleaner way to build printed text: **f-strings**. Put an `f` right before the quotation marks, and any variable name inside `{ }` gets swapped in automatically.

````python
mineral_name = "quartz"
density = 2.65

# old style
print("The density of", mineral_name, "is", density, "g/cm3")

# f-string style
print(f"The density of {mineral_name} is {density} g/cm3")
````
