### Why does this matter?

Every week, new course files will appear in your `course-materials/` folder. You'll download them to your ExploreHub through a tool called **nbgitpuller**. I am able to update that folder on GitHub and you are able to "pull" a copy of that folder to your ExploreHub using the buttons on the website.

Here's where it gets a bit complicated. Every time you press those  buttons, you'll pull a copy of the "upstream" version of `course-materials/` that I have updated. nbgitpuller doesn't just overwrite blindly. If you have a file in your ExploreHub with the same name as an updated file upstream, and you've made changes to it, nbgitpuller will preserve your version by renaming it with a timestamp — it won't be deleted. But that means working directly inside `course-materials/` can leave you with confusing duplicate, timestamped copies of files, and it's easy to lose track of which one has your actual work. 

If you have files in your ExploreHub that are named the same as files in the upstream `course-materials/` folder, they will be overwritten. 

That effectively means we shouldn't work *directly* inside `course-materials/`. Instead, we will always rename first and move our working copies into our personal folders. It keeps things clean and avoids all of that mess entirely.

So here's our simple rule:

> **Open from `course-materials/` → rename → copy to your personal folder → work there.**

Python is an **interpreted** language, which means that when a program is run, individual lines of code are translated into binary and executed one at at time. <u>There are several pros to interpreted languages:</u>
- programs tend to be faster to write
- source code *is* the program
- it's portable from one kind of computer to another so long as an interpreter is installed
<p><u>But there are some cons:</p></u>

- it requires an *interpreter* to run
- it's often solwer to execute than compiled code.

### What is an interpreter?
A **Python interpreter** is the program tha reads your Python code, line by line, and actually carries out the instructions. It translates what you wrote into something the computer can dd and gives you back a result.

This is in contrast to a *compiler* (which translates a whole program into machine code all at once before anything runs). 

Python will go one line at a time, which is part of why you can run a single cell and immediately see output, or get an error that stops right where the problem is.

### What is a Jupyter Notebook?

A Jupyter notebook (`.ipynb`) is a document that combines:
- **Written text and notes** (Markdown cells)
- **Python code** (Code cells)
- **Outputs** — results, plots, tables — right below the code that created them

It's like a lab notebook that can also run experiments.

#### To **create a new cell**, click the + button on the toolbar
  - Enter *edit mode* by double-clicking
  - Enter *view mode* by running the cell (> Run button, or type Shift+Return)


### Two cell types

| Cell type | What it does | How to run it |
|-----------|-------------|---------------|
| **Markdown** | Renders formatted text, headings, lists, equations | `Shift + Enter` → typesets the text |
| **Code** | Executes Python | `Shift + Enter` → runs the code, shows output below |

- You can change the type of cell with the dropdown menu on the toolbar
- You can delete cells by clicking into the cell and either pressing the scissor button in the tab menu, the trashcan in the upper right hand corner or by typing `dd`

Math Syntax Using $\LaTeX$

Markdown cells can render mathematical notation using $\LaTeX$ (pronounced "lay-tek"), a typesetting language built for equations.

### Inline vs. display math

| Style | Syntax | Renders as |
|---|---|---|
| Inline | `$v = d/t$` | appears within a line of text, like $v = d/t$ |
| Display | `$$v = d/t$$` | appears centered on its own line: $$v = d/t$$ |

Use inline math when a short expression fits naturally in a sentence, and display math to feature an equation on its own.

### Common $\LaTeX$ syntax

| Element | Syntax | Renders as |
|---|---|---|
| Fraction | `\frac{x}{y}` | $\frac{x}{y}$ |
| Exponent | `x^2` | $x^2$ |
| Subscript | `x_1` | $x_1$ |
| Square root | `\sqrt{x}` | $\sqrt{x}$ |
| Greek letters | `\pi`, `\alpha`, `\Delta` | $\pi$, $\alpha$, $\Delta$ |

Computers organize files in a **hierarchical file system** — a tree of folders (directories) and files.

```

└── home/
    └── susannah/              ← your home directory on ExploreHub

        │   └── week01/
        │       └── 01_course_intro.ipynb   ← this file!
```

The word **directory** means the same thing as **folder** — geologists say folder, computer scientists say directory. You'll see both.

## Part 5 — Paths

### What is a path?

Earlier, we talked briefly about paths. A **path** is the address that tells the computer exactly where a file or folder lives in the files sytem.

> Paths matter in data science!!

Everytime you load a dataset, you need to give Python the correct path.

### There are two kinds of paths:

**Absolute path** - the full address from the root of the file system:
```
/home/jupyter-susannah/Morey_EES3350/lectures/01_course_intro.ipynb
```

**Relative path** - the address *relative to where you currently are*:
If you're already in `/home/jupyter-susannah/` (i.e. `os.getcwd()` returns that path), then: 
```
Morey_EES3350/lectures/01_course_intro.ipynb
```
will point to the same file.

### Navigation shortcuts
If you're in a folder and need to acces files in another folder, you can use the following shortcuts:

| Shortcut | Meaning |
|----------|--------|
| `.` | Current directory |
| `..` | One level up |
| `../..` | Two levels up |
| `~` | Home directory |

So if you're in your personal home directory and you want to go grab some data from the `week01` folder in `course-materials`,  `..course-materials/week01/data.csv` means: "go up one folder, then into `week01/`, then grab `data.csv`."

### Bonus material (if time): `pathlib` module

It's sometimes helpful to store your home directory path as some variable. This makes it easier to access things later on. We can do that using the `pathlib` module.

````python
# first, let's import the module
from pathlib import Path
````

**We can combine `os` with `Path` to list the contents of your home directory**, even if you are not in that directory.

````python
os.listdir(Path.home())
````

**We can then build a path to a file using `pathlib`.** This is useful because path objects handle the `/` vs the `\` difference between Mac/Linux and Windows automatically.

````python
my_lectures = Path.home() / 'Morey_EES3350' / 'lectures'
print('Lectures folder:', my_lectures)
print('Does it exist?', my_lectures.exists())
````

### Why pathlib?

You can write paths as plain strings (`'folder/subfolder/file.csv'`), and for simple cases that's fine. But `pathlib.Path` is more powerful:

- Works on Windows, Mac, and Linux automatically (`/` vs `\`)
- Has useful methods like `.exists()`, `.stem`, `.suffix`, `.parent`
- Makes it easy to build paths programmatically with `/`

We'll use both styles — you should recognize either.

````python
# Anatomy of a path
example = Path('/home/jupyter-susannah/Morey_EES3350/lectures/01_course_intro.ipynb')

print('Full path:  ', example)
print('Parent dir: ', example.parent)
print('Filename:   ', example.name)
print('Stem (no ext):', example.stem)
print('Extension:  ', example.suffix)
````
