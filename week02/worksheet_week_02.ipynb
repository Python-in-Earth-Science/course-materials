## Wait, is that the same list?

Read carefully, run each cell, and think about what you *expect* to happen before you run it. 

Suppose two students each typed out the Nashville precipitation data by hand and you want to check whether they got the same values.

````python
list_a = [4.02, 4.47, 4.52, 4.72, 5.02, 4.36, 4.16, 3.79, 3.80, 3.36, 3.86, 4.43]
list_b = [4.02, 4.47, 4.52, 4.72, 5.02, 4.36, 4.16, 3.79, 3.80, 3.36, 3.86, 4.43]

print(list_a == list_b)
````

**Before you run the next cell, guess**: `list_a` and `list_b` contain the same values--we saw that in the above cell. Do you think `list_a` and `list_b` are the *same list*, as in the same object stored in teh same place in memory? Write your guess before running the cell below.

> your guess here.

````python
print(list_a is list_b)
````

- `==` checks whether two things have the **same value**
- `is` checks whether two things are **the exact same object**

`list_a` and `list_b` happen to hold identical values right now, but they're two separate lists sitting in two separate places, which is why `is` returns `False`.

#### Here's a bit of a tricky thing.

````python
list_c = list_a # NOT a copy! This just gives list_a a second name

# let's modify list_c
list_c[0] = 999 

print("list_c:", list_c)
print("list_a:", list_a)
````

<div class="alert alert-warning">

**Question:** `list_a[0]` changed even though you only touched `list_c`. Based on what you just saw, in your own words: what does the line `list_c = list_a` actually do? Why did modifying `list_c` also change `list_a`.

</div>

> your answer here

**The fix**: if you actually want an independent copy of a list (one you can change without affecting the original, use `.copy()`:
```python
list_d = list_a.copy()
```
Now `list_d` is a brand new list with the same values, not just another name for `list_a`.
