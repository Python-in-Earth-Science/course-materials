# Part 1 — Mutability and List Methods

Recall from Wednesday: lists are **mutable**, meaning you can change what's inside them after creating them. Let's see that in action.

## Changing a value

Say we made a typo-- Venus's radius should be 6052 km, not 6050. We can fix a single item in a list by assigning a new value to its index, the same way we'd assign a value to a variable.

```python
print(radius_km)
radius_km[1] = 6052
print(radius_km)
```

##### This only works because lists are mutable. Once we get to tuples later today, you'll see what happens when you try this on something that *isn't*.


## Adding to a list with `.append()`

Our lists only cover four terrestrial planets. What if we want to add Jupiter? `.append()` adds a new item to the **end** of an existing list.

```python
planet_names.append("Jupiter")
print(planet_names)
```

Notice how `.append()` uses **dot syntax**, just like `math.sqrt()`. That's becasue it's a **method** that belongs ot the list. Different data types have different methods available to them. We'll see more of these as the semester goes on.

### More list methods: `.copy()`, `.sort()`, `.insert()`, `.remove()`, `.pop()`

The `.append()` method isn't the only method that changes a list. Let's try a few more, but on a **copy** of `radius_km`, so we don't disturb the original list.

> #### `.copy()` makes an unlinked copy of a list
If you use `.copy()`, you'll be able to make a copy of a list that is completey disconnected to the original list

```python
radius_practice = radius_km.copy()
print(radius_practice)
```

> #### `.sort()` reorders a list in place, smallest to largest by default
If you use this method like we do below, `radius_practice` will now be sorted by size, but no longer lines up with `planet_names` in teh same order. This is another example of the parallell-lists problem from earlier: `.sort()` changes one list without knowing (or caring) about any others. Be careful sorting a list that is part of a parallel set.

```python
radius_practice.sort()
print(radius_practice)
```

> #### `.insert(index, value)` puts a new item at a specific position, rather than tacking it onto the end like like `.append()` does

```python
radius_practice.insert(0, 99999)   # insert a placeholder value at index 0
print(radius_practice)
```

> #### `.remove(value)` deletes the **first matching item it finds**.
Note: you give the value itself, not an index

```python
radius_practice.remove(99999)
print(radius_practice)
```

> #### `.pop(index)` removes an item at a given index **and hands it back to you**
THis is useful when you want to both remove something and keep it elsewhere

```python
last_item = radius_practice.pop(-1)
print("Removed:", last_item)
print(radius_practice)
```

## Part 2 — Nested Lists

Right now we have four separate lists that only stay connected because we're careful to keep them in the same order. There's a more compact way to bundle related data together: a **list of lists**, also called a **nested list**.

```python
planet_data = [
    ["Mercury", 9083, 0.068, 2440],
    ["Venus", 2636, 0.77, 6052], 
    ["Earth", 1360, 0.31, 6370],
    ["Mars", 586, 0.25, 3390],
    ["Jupiter", 50.5, 0.343, 69911]
]

print(planet_data)
```

### What did we just do?

Each item in `planet_data` is itself a list-- one planet's full record `[name, solar_irradiance, albedo, radius_km]`. To get the full record for a planet, we can index into the outermost list.

What do I mean by "outermost list"? When we create nested lists, there will be an outermost list and an innermost list.
- The **outermost list** is the `planet_data` instelf--teh list that holds everything else. Indexing into it once, like `planet_data[0]`, hands you back one entire inner list.
- The **innermost list** is each individual planet's record, like `["Mercury", 9083, 0.068, 2440]`. INdexing into *that* (the second set of brackets) gets you down to a single value.

### Indexing into a nested list

Here's how that looks.

We can grab Earth's name (the 0th index of an inner list which sits in the 2nd index of the outer list) by calling:

```python
planet_data[2][0]
```

To index into nested lists:
- the **first set of brackets** will reference the outer list
- the **second set of brackets** will reference the inner list

<div class="alert alert-success">

**Tip:** Think of it as narrowing down in two steps: `planet_data[2]` gets you *to* Earth's list, and the second `[1]` gets you *into* that list to a specific value. This is sometimes called **matrix-style indexing**, since it works like rows and columns.
</div>

## Part 3 — Tuples

Lists aren't the only way to group values together. A **tuple** looks almost identical to a list, but with one key difference — it's **immutable**: once created, it can't be changed.

### Creating a Tuple
The synax is nearly the same as a list. The only difference is parentesis `()` instead of square brackets `[]`.

```python
terrestrial_planets = ("Mercury", "Venus", "Earth", "Mars")
print(terrestrial_planets)
print(type(terrestrial_planets))
```

Tuples support indexing and slicing exactly the way lists do.

```python
print(terrestrial_planets[0])
print(terrestrial_planets[1:3])
```

### What happens if you try to change one?

Let's try the same index-assignment trick we used on `radius_km` earlier, but on a tuple this time.

```python
terrestrial_planets[0] = "Not Mercury"  # this will raise an error -- that's expected!
```

<div class="alert alert-success">

**Tip:** That's a `TypeError` — a new error type. It means you tried to do something with an object that its type doesn't support. **Tuples don't support item assignment; lists do.** That's the whole point of a tuple: it's a guarantee that the data won't change by accident.
</div>

### A real-world use case: coordinates
One of the msot common places you'll see tuples in scientific Python is a pair of coordinates. A location's latitude and longitude shouldn't accidentally change partway through your analysis, which makes a tuple a natural fit.

```python
nashville_coords = (36.1627, -86.7816)   # (latitude, longitude)
print(nashville_coords)
print("Latitude:", nashville_coords[0])
print("Longitude:", nashville_coords[1])
```

You'll see this exact pattern again in Unit 3, when we start working with real geospatial data — station locations, sample sites, and grid coordinates are almost always stored as immutable pairs like this.
