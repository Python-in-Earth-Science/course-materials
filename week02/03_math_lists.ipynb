### What's up with the `.` you put after `math`?

Recall from Lecture 1 that in Python, everything is an **object** and every object carries it's own **attributes** (data attached it, sometimes called metadata) and **methods** (functions attached to it). the dot (`.`) is how you reach into an object and grab one of those.

### A function *you* write looks like this
The `math` functions above were written by someone else. But starting next week, you'll be writing your own. And they're not as mysterious as they might seem! 

**Remember how many times you did the same calculation in the lab?** Instead of repeating the WHOLE calculation each time, we can write a function to quickly do this calculation whenver we want. Here's teh exact blackbody temperature calculation from Lab 1 (Eq. 8, with albedo), wrapped up as a function:

````python
def blackbody_temp(S, a, stef_boltz=5.67e-8):
    """
    Predict a planet's blackbody surface temperature (K) from its solar constant (S) and albedo (a).
    """
    T = ((1 - a) * S / (4 * stef_boltz)) ** (1/4)
    return T
````

- `def` starts a function definition, followed by a name you choose (`blackbody_temp`) and the **parameters** it needs in parentheses — the inputs the function expects each time you use it.
- Everything indented underneath is the function's *body* — the steps it runs.
- `return` sends a value back out, the same way `math.sqrt(16)` hands you back `4.0`.

For now, don't worry about writing your own — just notice that this is nothing more than Eq. 8 from Lab 1, given a name.

````python
T_earth = blackbody_temp(S=1372, a=0.31)
T_mars = blackbody_temp(S=586, a=0.25)

print(T_earth)
print(T_mars)
````

# Part 2 — Lists
So far, every variable you've created has stored a single value. What if you wanted to store several related values under one name, like the solar irradience of four different planets?
> **Lists** are Python's way of storing multiple values in a single variable. They are defined with square brackets, with commas between items: `a = [4.-, 2.5, 5.1]`. Lists are **mutable**, meaning you can change what's inside them after you create them.

We are going to use planet data for the rest of this lecture, pulling in Mercury and Venus alongside Earth and Mars values from Lab 1.

| Planet      | Solar irradiance (W/m²) | Albedo | Radius (km) |
|-------------|--------------------------|--------|-------------|
| Mercury     | 9,083                    | 0.068  | 2,440       |
| Venus       | 2,636                    | 0.77   | 6,050       |
| Earth       | 1,360                    | 0.31   | 6,370       |
| Mars        | 586                      | 0.25   | 3,390       |

We can create lists of this data like this:

````python
planet_names = ["Mercury", "Venus", "Earth", "Mars"]
solar_irradiance = [9083, 2636, 1360, 586] # W/m^2
albedo = [0.068, 0.77, 0.31, 0.25]
radius_km = [2440, 6050, 6370, 3390]
````

And then we can print the whole list like this:

````python
print(planet_names)
````

What if we want to access one item from that list? that's where indexing comes in...

### Indexing

In Python, the **first item in a list has index 0**, not 1. To get an item out of a list, put its index in square brackets after the list name.

````python
print(planet_names[0]) # first planet in list has index 0
print(planet_names[1]) # second planet in list has index 1
````

<div class="alert alert-success">

**Tip:** **This is what we call "`zero indexing`".** Zero-based indexing feels backwards at first — just remember that `list[0]` means "the first step from the start," not "item number one." It becomes automatic with practice.
</div>

### Negative Indexing
You can also index from the **end** of a list using negative numbers. `list[-1]` is always the last item, no matter how long the list is.

````python
print(planet_names[-1]) # last planet
print(planet_names[-2]) # second to last planet
````

<div class="alert alert-success">

**Tip:** Negative indexing is especially handy when you don't know (or don't want to count) how long a list is — `list[-1]` always works, whether the list has 4 items or 4,000.
</div>
