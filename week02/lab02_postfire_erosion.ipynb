#### Wildfires are increasing wtih climate change
Wildfires are a yearly occurence for the Western US, with 1000's of fires sparking most commonly during August through November. Dry vegetation and lack of precipitation in summer months enhance fire vulnerability. It has been shown over the past few decades that a warming climate lengthens the wildfire season (Dennison et al., 2014). Fire activity not only affects the organisms (including humans) that inhabit or use this land, but also impacts the post-fire geomorphological processes and sediment transport processes on hillslopes that burned.

![Increase of wildfires in the western US](../figures/IncreaseofwildfiresUSWest.png)

<span style="font-size:0.85em;color:#6b7a8d;"><em>Denison et al. (2014)</em></span>

#### Wildfires can make landscapes more prone to erosion
There are two main changes that wildfires make to to landscapes that we will focus on in this lab:
##### 1. Wildfires remove vegetation.
Losing vegetation directly affects thye hydologic cycle: with teh canopy gone, less rainfall is intercepted before it reaches the ground. The soil will also lose the protection that plants provided from wind and rain.

##### 2. Wildfires change soil properties.
Soil that has experienced fire will undergo physical and chemical changes to the structure of the soil. Ash from a fire will affect the pore space between grains, reducing water infiltration rates. Also, it turns out that experiencing heat will cause a waxy coating to form on soil grains (baked out of burning organic matter). Both of these things combine to enhance **soil water repellency**. This means water will not easily infiltrate soils that have recently experienced fire.

![burnseverity-soilrepellency.png](../figures/burnseverity-soilrepellency.png)


#### How do these changes affect the Water Balance Equation?
The water balance in its simplest liunear form is:
$$P = Q+ET+\Delta S$$
where
- $P$ = precipitation in mm/time</p>
- $Q$ = runoff in mm/time</p>
- $ET$ = evapotranspiration in mm/time </p>
- $\Delta S$ = the change in groundwater storage due to infiltration in mm/time</p>

When infiltration drops ($\Delta S$) after a fire, more of that incoming precipitation has to leave as runoff ($Q$Q) instead, increasing the risk of:
- shallow landslides
- debris flows
- downstream flash flooding

Large-scale debris flows in particular can be extremely destructive and can reshape a landscape rapidly (Santi and Rengers, 2020)

#### Debris Flow Hazards
A debris flow is a moving mass of loos mud,s and, soil, rock, water and air that travels down a slope under the influence of gravity. To be considered a debris flow, the moving material must be loose and capable of "flow", and at least 50% of the material msut be sand-sized particles or larger.

Because debris flows travel relatively fast and carry a ton of mass, they are very destructive. Debris flows are a major hazard in the years directly after a fire. In the Western US, the US Geological Society (USGS) is mandated to make post-fire risk maps that show the liklihood of dbris flow occurence for different streams.

##### Here is the debris flow hazard map constructed after the Calwood Fire.

![Calwood fire debris flow hazard](../figures/calwood_extent.png)

This USGS Debris Flow Hazard Map was published in October 2020, immediately after the Calwood Fire had been contained. It shows the likelihood of debris flow occurence determined by:
- terrain steepness
- stream order
- burn severity
- predicted future rain fall intensity (the trigger to actually initiate a debris flow)

**Dark red colors indicate a higher risk of debris flow while yellow colors indicate a lower risk of debris flows.**

# VI. The Cane (1980) Debris-Flow Threshold
The first-generation empirical debris-flow occurance model was formulated by Cain (1980). It found that debris flows occur in association with high-intensity, short-duration rain events, following a threshold:
$$I = 14.82D^{-0.39}$$
where $I$ is the threshold rainfall intensity (mm/hr) and $D$ is the storm duration (hrs). If the *actual* rainfall intensity during a storm meets or exceeds $I$ for that storm's duration, the model predicts a debris flow is likely. Graphically, this looks like this:

![caine data](../figures/caine_figure.png)

#### Imagine three storms hit the Calwood burn area:

|storm|intensity (mm/hr) | duration (hr)
|---|:---:|:----:|
| storm 1 | 30 | 1 |
| storm 2 | 18 | 0.2 |
| storm 3 | 45 | 3 |

<div class="alert alert-info">

### 18.

**Represent each storm as a tuple of `(intensity_mm_per_hr, duration_hr)`.**

For each storm **calculate the Caine (1980) threshold intensity for that storm's duration**, then determine whether the storm's actual intensity meets or exceeds the threshold (a boolean).

**Print a sentence for each storm stating the threshold, the actual intensity, and whether a debris flow would be predicted.**

    
</div>

````python
# your answer
````

````python
storm1 = (30,1)
storm2 = (18, 0.2)
storm3 = (45, 3)

# storm 1
I1, D1 = storm1
threshold1 = 14.82*D1**(-0.39)
debrisflow1 = I1 >= threshold1
print(f"Storm 1: intensity = {I1} mm/hr, duration {D1}, threshold = {threshold1:.2f} mm/hr --> debris flow predicted? {debrisflow1}")

# storm 2
I2, D2 = storm2
threshold2 = 14.82*D2**(-0.39)
debrisflow2 = I2 >= threshold2
print(f"Storm 2: intensity = {I2} mm/hr, duration {D2}, threshold = {threshold2:.2f} mm/hr --> debris flow predicted? {debrisflow2}")

# storm 3
I3, D3 = storm3
threshold3 = 14.82*D3**(-0.39)
debrisflow3 = I3 >= threshold3
print(f"Storm 3: intensity = {I3} mm/hr, duration {D3}, threshold = {threshold3:.2f} mm/hr --> debris flow predicted? {debrisflow3}")
````

**A real historic storm.** Rainfall on Sept 11-13, 2013 triggered over 11,138 debris flows across a 3,430 km$^2$ area of the Colorado Front Range, including areas very close to what would become the Calwood Fire perimeter seven years later. The storm had two especially intense rainfall periods: a 12.5-hour period with a maximum 10-minute intensity of 67 mm/hr, and an 8-hour period with a maximum 10-minute intensity of 39 mm/hr (Coe et al., 2016).

**Represent each of these two rainfall periods as a tuple** of `(intensity_mm_per_hr, duration_hr)`, then apply the Caine (1980) threshold model to each, eactly as you did in Question 18. Print your results.

**Then, in a Markdown cell, briefly discuss:** does this model predict a debris flow for both periods? Does that match what actually happened during the 2013 Event?

````python
# your answer here
````

````python
period1_2013 = (67, 12.5)
period2_2013 = (39, 8)

I_p1, D_p1 = period1_2013
threshold_p1 = 14.82 * D_p1**(-0.39)
debrisflow_p1 = I_p1 >= threshold_p1
print(f"2013 Event Period 1: intensity = {I_p1} mm/hr, duration = {D_p1} hr, threshold = {threshold_p1:.2f} mm/hr --> debris flow predicted? {debrisflow_p1}")


I_p2, D_p2 = period2_2013
threshold_p2 = 14.82 * D_p2**(-0.39)
debrisflow_p2 = I_p2 >= threshold_p2
print(f"2013 Event Period 2: intensity = {I_p2} mm/hr, duration = {D_p2} hr, threshold = {threshold_p2:.2f} mm/hr --> debris flow predicted? {debrisflow_p2}")
````

The model predicts a debris flow for both rainfall periods, since the actual intensity comfortably exceeds the Caine threshold in each case. This matches the real outcome: the 2013 Event triggered over a thousand debris flows across the northern Front Range, making it one of the most significant debris-flow events on record for the region (Coe et al., 2016). It's worth noting that the Caine threshold was developed form a global dataset of many different storms and terrains, so it won't predict every event perfectly, but here it correctly flags a storm we know, from the historical record, did trigger widespread debris flows.

**Conceptual question.** We built two different structures for the same underlying information in this lab: `gully_all_data` (a nested *list*) and `Calwood_Fire_gullies` (a nested *dictionary*). Both store a name, an area, and a burn severity for each gully.

In a Markdown cell, explain: **for a dataset like this one, what's an advantage of using a nested dictionary over a nested list?**


````python
# your answer here
````

With `gully_all_data`, you ahve to know a gully's position in the list to retrieve its record. `gully_all_data[2]` only gets you Geer Canyon's data because you (or the code that built the list) remembered that Geer Canyon was the third gully added. If someone reorders the list or ads a new gully in the middle, every hard-coded index elsewhere in the code silently points to the wrong gully.

With `Calwood_Fire_gullies`, you retreive a gully's record by its name regardless of what order the dictionary's keys happen to be in internally. That makes dictionaries a better fit whenver the natural way to look something up is "by name" rather than "by position", which is exactly the case for a set of named field sites.
