### Creating a dictionary
Dictionaries use curly braces `{}`. Each entry is a **key** linked to its **value** with a colon `:`, and entries are separated by commas. We call these **key-value pairs**.

#### Let's create a dictionary that stores all the information for Earth:

````python
earth_dict = {
    "name": "Earth",
    "solar_irradiance": 1360,
    "albedo": 0.31,
    "radius_km": 6370
}

print(earth_dict)
````

##### Compare this to `planet_data[2]`, which stores the exact same information: `["Earth", 1360, 0.31, 6370]`. Our dictionary has the same data, but now every value comes with a label isntead of a position you have to remember.

### Accessing a value by key
Instead of an index number in square brackets, you will use the key.

````python
print(earth_dict["solar_irradiance"])
print(earth_dict["albedo"])
````

What would happen if you ask for a key that doesn't exist?

````python
print(earth_dict["radius"])
````

### A second use for dictionaries: mapping one thing to another
Dictionaries aren't just for storing one objects attributes. They're also useful any time you want to look something up by name. Here's each planet mapped to its number of known moons:

````python
moons = {
    "Mercury": 0,
    "Venus": 0,
    "Earth": 1,
    "Mars": 2,
    "Jupiter": 95
}

print(moons["Jupiter"])
````

#### **NOTE**: in a dictionary, the **keys are immutable**. You can't rename "Jupiter" to something else, the same way you can't reassign an item in a tuple. But the **values are mutable**. So `moons["Jupyter"]=96` works jsut fine.

Moral of the story: keys and values follow different rules.

### Adding a new key-value pair
On Friday, we appended Saturn to `planet_data`. Let's add it here too. Adding a new key-value pair uses the exact same syntax as accessing one, just with an assignment.

````python
moons["Saturn"] = 146
print(moons)
````

### Updating an existing value
That same syntaxis also works to correct a value that's already there. Python doesn't need you to tell it whether you're adding or updating, it just checks whether the key already exists.

````python
print(moons)
moons["Earth"]=7
print(moons)
````

## Part 2 — Nested Dictionaries

We just built `earth_dict` and `mercury_dict` separately. What if we want *all* of the planets' full records in one place, the same way `planet_data` held all of them as a nested list? We can do the same thing with dictionaries: let's create a dictionary whos values are themselves dictionaries.

````python
planets = {
    "Mercury": {"solar_irradiance":9083, "albedo":0.068, "radius_km":2440},
    "Venus": {"solar_irradiance": 2636, "albedo":0.77, "radius_km":6052},
    "Earth": {"solar_irradiance": 1360, "albedo":0.31, "radius_km":6370},
    "Mars": {"solar_irradiance": 586, "albedo":0.25, "radius_km":3390},
    "Jupiter": {"solar_irradiance": 50.5, "albedo":0.343, "radius_km":69911}
}

print(planets["Earth"])
````

#### In this case:
The outer keys are planet names where each one points to *another dictionary* of attributes. This is the dictionary equivalent of nested lists from Friday and it's exactly the structure you'll see in real-world config files and JSON data later in the semester.

To reach a specific attribute, chain two keys together:
- **The first** picks the planet
- **The second** picks the attribute

This is the same tow-step logic as `planet_data[2][1]` from Friday, jsut with labels instead of positions.

````python
def blackbody_temp(S, a, stef_boltz=5.67e-8):
    """Predict a planet's blackbody surface temperature (K) from its solar constant (S) and albedo (a)."""
    T = ((1 - a) * S / (4 * stef_boltz)) ** (1/4)
    return T
````

### Checking a key safely: `.get()` and `in`
Earlier, asking for a key that doesn't exist raised a `KeyError`. Sometimes that's exactly what you want: a loud failure that tells you something's wrong. Other times, you'd rather Python just tell you the key isn't there, without crashing. To do that, we can just:

````python
print(planets.get("Saturn"))
````

##### `.get()` will return `None` instead of raising an error when a key is missing.
> **What is `None`, exactly?**
`None` is its own special Python type (`NoneType`), used to represent "nothing here." It isn't `0`, it isn't an empty string `""`, and it isn't `False` — it's Python's explicit way of saying a value doesn't exist. You'll see `None` again throughout the semester any time something comes up empty.

You can also give `.get()` a default value to return instead:
```python
planets.get("Saturn", "not in dictionary")
```

````python
print(planets.get("Saturn", "not in dictionary"))
````

##### You can also check whether a key exists *before* trying to access it using the `in` operator. It will return a bool.

````python
print("Saturn" in planets)
print("Earth" in planets)
````

## Part 3 — Which structure should I use?

This week you've learned three ways to organize more than one value: lists, tuples, and dictionaries. Here's a quick way to think about choosing between them:

| Structure | Ordered? | Mutable? | Access by | Good for |
|---|---|---|---|---|
| **List** | Yes | Yes | Position (index) | A collection of similar items that might change |
| **Tuple** | Yes | No | Position (index) | Fixed data that shouldn't change, like coordinates |
| **Dictionary** | Yes* | Values only | Key (label) | Labeled data, or looking something up by name |

*Dictionaries do remember the order you added things in Python, but you should never *rely* on that order the way you can with a list's position.

## Extra Material (if time)
Three more operators are worth having in your toolkit:
- **membership** operators for asking "is this in that"
- **logical** operators for combining multiple `True`/`False` checks into one
- **identity** operators for checking whether two thigns are literally the **same object** in memory *not just equal in value*)

### Membership operators : `in` and `not in`
You already used `in` a few minutes ago to check whether a key exists in a dictionary. That same operator works identically on strings, lists, and tuples. Basically it's used anywhere Python chan check: "is this item inside that collection." `not in` just flips the answer.


````python
planet_names = ["Mercury", "Venus", "Earth", "Mars", "Jupyter"]

print("Earth" in planet_names)
print("Saturn" in planet_names)
print("Saturn" not in planet_names)
````

### Logical operators : `and`, `or`, `not`
Logical operaters let you combine multiple `bool` checks into a single `True`/`False` answer, without needing an `if` statment (don't worry, we'll get there soon). 
-`and` requires both sides to be `True`
- `or` only needs one side to be `True`
- `not` flips a single blool

````python
print(moons["Earth"] > 0 and moons["Mars"] > 0)   # both planets have at least one moon
print(moons["Mercury"] > 0 or moons["Venus"] > 0) # neither planet has any moons
print(not (moons["Mercury"] > 0))                 # flips the Mercury check in the line above
````

### Identity operatros : `is` and `is not`
`==` checks whether two values are *equal*. `is` checks something stricter: whether two names point to the *exact same object* in memory. For most values these give the same answer, but for `None` specifically, `is` is the correct tool to use (because there's only ever **one** `None` object in a running Python program. `saturn == None` will technically work, but `saturn_lookup is None`is the convention you'll see (and should use) everywhere.

````python
saturn_lookup = planets.get("Saturn") # Saturn was never added to planets, so this is None

print(saturn_lookup == None)
print(saturn_lookup is None)
````
