## Part 0 : Picking up where we left off

````python
gully_names = ["Heil Ranch Gully 1", "Heil Ranch Gully 2", "Geer Canyon", "Central Gulch"]
normal_infiltration = [16, 16, 22, 16]
postfire_infiltration = [2.25, 11.77, 3.5, 4.0]
drainage_area = [0.1538, 0.2523, 7.325, 0.1515]
burn_severity = ["severe", "moderate", "severe", "moderate"]

def runoff_rate(P, ET, S):
    """Calculate a storm's runoff rate from the water balance equation."""
    Q = P - ET - S
    return Q

def debris_flow_threshold(D):
    """Calculate the Caine (1980) rainfall-intensity threshold for debris flows."""
    threshold = 14.82 * D ** (-0.39)
    return threshold

def check_debris_flow(I, D):
    """
    Determine whether a storm is predicted to trigger a debris flow using the
    Caine (1980) intensity-duration threshold.

    Parameters
    ----------
    I : float
        storm rainfall intensity (mm/hr)
    D : float
        storm duration (hr)

    Returns
    -------
    threshold : float
        Caine (1980) threshold intensity (mm/hr)
    debris_flow : bool
        whether a debris flow is predicted (I >= threshold)
    """
    threshold = 14.82 * D ** (-0.39)
    debris_flow = I >= threshold
    return threshold, debris_flow

# just a test to be sure it works...
threshold, debris_flow = check_debris_flow(I=6, D=12.5)
print(threshold, debris_flow)
````

## Branching (making decisions in code)
Think of branching like making a decision:
- **IF** it's raining, I'll take a rain coat.
- **ELSE**, I won't!

Code can "decide" what to do next based on whether a condition is `True` or `False`. This is called `branching`. Today, we'll build up from the simpliest form (`if`) to combining multiple conditions (`and`/`or`/`not`) and nesting decisions inside other decisions, all usig the gully data set you already know from Lab 02.

## Part 1 : The `if` statement
This is the simplest form of branching. Lat time, `check_debris_flow()` handed us a `debris_flow` boolean. Now let's actually do something with it.

Previously, we dealt with the outpt from `check_debris_flow` like this:

````python
threshold, debris_flow = check_debris_flow(I=6, D=12.5)
print(f"With a peak intensity of 6 mm/hr, a debris flow is predicted: {debris_flow}")
````

##### That's kind of awkward, right?

`debris_flow` is a boolean so when we print it, it will always be either `True` or `False.

#### `if` statments do something with that information. They have the ability to do something with the boolean.

````python
threshold, debris_flow = check_debris_flow(I=6, D=12.5)

if debris_flow:
    print("Debris flow predicted for this storm. Debris flow predicted.")
````

Let's breakdown what happened:
| Piece | What it means | In cell above |
|---|---|---|
| **`if`** | keyword that starts the decision (turns green in Jupyter, just like `def` did for functions | `if` |
| **condition** | whatever follows `if` | `debris_flow` variable (which holds a boolean) |
| **colon** | tells Python that the indented code below belongs to this decision | -- |
|**indented content**|what to do if the condiiton is met | `print(""Debris flow predicted for this storm. Debris flow predicted.")`|

The condition can be a full comparison expression, or (as here) a variable that already holds a `bool`.

## Part 2 : The `if`-`else` statement
What if we want something to happen when the condition is `False` too'? Just add one more keyword:

````python
threshold, debris_flow = check_debris_flow(I=6, D=12.5)

if debris_flow:
    print("Runoff exceeds infiltration at Central Gulch. Debris flow predicted.")
else:
    print("No debris flow predicted for this storm.")
````

#### To use `else`:
- new line after the `if` block
- type `else:` (don't forget the colon!)
    - this should auto **un-indent**, but if it doesn't, do it yourself.
- new, line (should be auto **indented**, but if not, do it yourself), and write the code for the "otherwise" case

## Part 3 : The `if`-`elif`-`else` statement
What if there are more than two possibilities? One more keyword: `elif` (short for "else if") goes beetween `if` and `else`, and works the same way as `if`.

Instead of just a Pass/Fail (P/F) debris flow flag, let's classify **how much a gully's infiltration capcity dropped** after the fire, something that differs gully by gully.

````python
def classify_infiltration_recovery(normal, postfire):
    """Classify how severly a gully's infiltration capacity dropped after a fire."""
    if postfire < 0.2*normal:
        return "severe reduction"
    elif postfire < 0.6*normal:
        return "moderate reduction"
    else:
        return "minor reduction"
````

Let's test it using Central Gulch's change.

````python
result = classify_infiltration_recovery(normal_infiltration[3], postfire_infiltration[3])
print("Central Gulch:", result)
````

#### Common mistakes to avoid:
- forgetting the colon `:` after a condition
- not indenting the code inside an `if`/`elif`/`else` block
- using `=` (assignment) instead of `==` comparison
#### Some General Rules:
- Only want something to happen under one condition: **Use just `if`**.
- Want one thing *or* another thing, depending on a condition? **Use `if`/`else`**.
- Have several distinct possibilities? **Use `if`/`elif`/`else`**.

## Part 4 : Combining branching with other keywords
Sometimes a decision depends on **more than one condition** at once. `and` and `or` let you combine conditions into a single expression.

#### `and` : use when **both** conditions must be true

````python
n = 2
name = gully_names[n]

if drainage_area[n] > 1 and postfire_infiltration[n] < 5:
    print(name, "is a large gully with low postfire infiltration : HIGH PRIORITY.")
else:
    print(name, "does not meet both conditions.")
````

#### `or` : use when **either** condition can be true

````python
n=0
name = gully_names[n]

if drainage_area[n] > 1 or postfire_infiltration[n] < 3:
    print(name, "meets at least one monitoring criterion.")
else:
    print(name, "does not currently need priority monitoring.")
````

## Part 5 : nested `if` statements
You can put an `if` statement *inside* another `if` statement's block, using indentation. This is useful when a second decision only makes sense after the first condition is already true.

````python
n = 2
name = gully_names[n]

severity = burn_severity[n]
recovery = classify_infiltration_recovery(normal_infiltration[n],
                                           postfire_infiltration[n])

if severity == "severe":
    print(name, "experienced severe burn.")
    if recovery == "severe reduction":
        print("  ...and infiltration capacity dropped severely too -- highest concern.")
    else:
        print("  ...but infiltration capacity held up better than expected.")
else:
    print(name, "experienced moderate burn.")
````

<div class="alert alert-success">

**Tip**: Each level of nesting adds one more level of indentation. If you find yourself nesting more than 2-3 levels deep, it's often a sign the logic could be reorganized more clearly with `elif` or a combined `and`/`or` condition instead.

</div>

## Extra if Time: shorthadn you'll see in other people's code


### The ternary operator (`if`-`else` shorthand)
For simple `if`-`else` assignments, Python has a one-line shortcut:
```python
variable = value_if_true if condition else value_if_false
```
Let's compare using the long way and then this short way.
##### The long way:

````python
n=0

if burn_severity[n] == "severe":
    priority = "high"
else:
    priority = "standard"
````

##### The short way:

````python
priority = "high" if burn_severity[n] == "severe" else "standard"
````

Both do exactly the same thing. The shorthand is just more compact. You're not required to use it, but you'll see it in other people's code so it's worth being able to read.

### The `pass` keyword
