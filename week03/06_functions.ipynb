## Part 1 : The anatomy of a function

#### Let's try rebuliding `blackbody_temp()` together
You've seen this function twice now: during the first week of class as a preview and then again on Monday when we used it with dictionary data. Let's rebuild it together one more time, and this time slow down on every piece.

```python
def blackbody_temp(S, a, stef_boltz=5.67e-8):
    """
    Predict a planet's blackbody surface temprature (K) from its solar constant (S) and albedo (a).
    """
    T = ((1-a) * S / (4 * stef_boltz)) ** (1/4)
    return T
```

Now we can use `blackbody_temp` with our dictionary as inputs:

```python
T_earth = blackbody_temp(S = planets["Earth"]["solar_irradiance"], a = planets["Earth"]["albedo"])
print(T_earth)
```

| Piece | What it means | In `blackbody_temp` |
|---|---|---|
| `def` | keyword that starts a function definition | `def` |
| function name | what you'll write to call/run it later | `blackbody_temp` |
| parameters | placeholder inputs, listed in parentheses | `S`, `a`, `stef_boltz` |
| default value | a parameter's fallback value if you don't provide one | `stef_boltz=5.67e-8` |
| `:` | ends the definition line, same as `for` or `if` will do later | — |
| body | the indented block that runs every time the function is called | the calculation of `T` |
| `return` | sends a value back out to whatever called the function | `return T` |

## Part 2 : Calling a function: positional vs. keyword arguments
When you call a function, you hand values to its parameters. There are two ways to do that:
- **positional arguments** : Python matches your values to parameters **by their order** in the function definition
- **keyword arguments** : you explicity write `parameter=value`, so order doesn't matter

Both work, but keyword arguments are usually clearer to read, especially once a function has more than one or two parameters.

##### **Positional Use:**

```python
T_mars_positional = blackbody_temp(586, 0.25)
print(T_mars_positional)
```

##### **Keyword Use:**

```python
T_mars_keyword = blackbody_temp(a=0.25, S=586)
print(T_mars_keyword)
```

<div class="alert alert-success">

**Tip!:** You can mix styles--positoinal arguments first, then keywords arguments, but a positional argument can never come *after* a keyword argument. Python will raise a `SyntaxError` if you try.
    
</div>

Every function you've used so far (even `math.sqrt()`) started as someone typing out a `def` block like the one above. Let's write one completely from scratch, using something you already calculated by hand in Lab 02: the storm runoff rate from the water balance equation.
$$Q = P - ET - S$$

```python
def runoff_rate(P, ET, S):
    """Calculate a storm's runoff rate from the water balance equation."""
    Q = P - ET - S
    return Q
```

## Part 4 : Documenting your functions: docstrings and `help()`
You'e been writing one-line docstrings so far. For functions with multiple parameters and a return value, it's worth being more explicit. A common convention lists every parameter and what gets returned:

```python
def check_debris_flow(I, D):
    """
    Determine whether a storm is predicted to trigger a debris flow using the 
    Caine (1980) intensity-duration threshold.

    Parameters
    ----------
    I : float
        storm rainfall intensity (mm/hr)
    D : float
        storm duration (hr)

    Returns
    ---------
    threshold : float
        Cain (1980) threshold intensity (mm/hr)
    debris_flow : bool
        whether a debris flow is predicted (I >= threshold)
    """
    threshold = 14.82 * D**(-0.39)
    debris_flow = I >= threshold
    return threshold, debris_flow
```

#### `help`
Once a function has a docstring, Python can display it for you (or anyone else using your code) with the built-in `help()` function. It's the exact same tool you can use on functions you didn't write.

```python
help(check_debris_flow)
```

```python
help(math.sqrt)
```

## Part 5 : local vs global scope
You've just written several functions that calculate values like `threshold` and `Q` inside their own body. Variables created **inside** a function only exist inside that function. This is called **local scope**. Once the function finishes running, those variables disappear. Let's watch this cause an error on purpose. 

```python
def add_one(x):
    result = x + 1
    return result

print(add_one(5))
print(result)
```

## Part 6 : functions that plot
We've now calculated several numbers per gully-- infiiltration rates, runoff, debris-flow thresholds. Wouldn't it be nice to **see** one of them?

**First look**: We haven't formally covered plotting yet (that's coming in a few weeks with `matplotlib`), but I wanted to show you how we could incorporate plotting into a function. For today, just focus on the **pattern**: a plotting function takes in some data and, instead of `return`ing a value, it draws a figure. You'll see this exact library again soon.

```python
import matplotlib.pyplot as plt
```

#### Let's make the function `plot_infiltration_change()`
We want a function that takes one gully's name, it's infiltration data, and draws a simple bar chart comparing them.

```python
def plot_infiltration_change(gully_name, normal_rate, postfire_rate):
    """
    Plot a gully's infiltration rate before and after the fire.

    Parameters
    ----------
    gully_name : str
        Name of the gully.
    normal_rate : float
        Pre-fire infiltration rate (mm/hr).
    postfire_rate : float
        Post-fire infiltration rate (mm/hr).

    Returns
    -------
    None
    """
    plt.bar(["normal", "post-fire"], [normal_rate, postfire_rate], color=["#4c72b0", "#c44e52"])
    plt.ylabel("Infiltration rate (mm/hr)")
    plt.title(f"infiltration before vs. after fire")
    plt.show()

```

```python
name = gully_data["Geer Canyon"]
input1 = gully_data["Geer Canyon"]["normal_infiltration"]
input2 = gully_data["Geer Canyon"]["postfire_infiltration"]


plot_infiltration_change(name, input1, input2)
```

Notice `plot_infiltration_change()` has no `return` statement. It doesn't need one, since its whole job is to draw a figure, not hand back a value. Without this function, you'd have to copy-paste those same four lines once for every gully. This is the same motivation for writing a function that as before, just applied to a plot instead of a calculation.

#### Another example of a plotting function

```python
def plot_infiltration_all(gully_data):
    """
    Plot infiltration rate before and after fire for all gullies as a grouped bar chart.

    Parameters
    ----------
    gully_data : dict
        Dictionary keyed by gully name, where each value is a dictionary with
        "normal_infiltration" and "postfire_infiltration" keys (mm/hr).

    Returns
    -------
    None
    """
    names = list(gully_data.keys())
    normal_rates = [gully_data[names[0]]["normal_infiltration"],
                     gully_data[names[1]]["normal_infiltration"],
                     gully_data[names[2]]["normal_infiltration"],
                     gully_data[names[3]]["normal_infiltration"]]
    postfire_rates = [gully_data[names[0]]["postfire_infiltration"],
                       gully_data[names[1]]["postfire_infiltration"],
                       gully_data[names[2]]["postfire_infiltration"],
                       gully_data[names[3]]["postfire_infiltration"]]

    width = 0.35
    x_normal = [0, 1, 2, 3]
    x_postfire = [0.35, 1.35, 2.35, 3.35]
    x_ticks = [0.175, 1.175, 2.175, 3.175]

    plt.bar(x_normal, normal_rates, width=width, label="normal", color="#4c72b0")
    plt.bar(x_postfire, postfire_rates, width=width, label="post-fire", color="#c44e52")

    plt.xticks(x_ticks, names)
    plt.ylabel("Infiltration rate (mm/hr)")
    plt.title("Infiltration rate before vs. after fire, by gully")
    plt.legend()
    plt.show()

plot_infiltration_all(gully_data)
```
